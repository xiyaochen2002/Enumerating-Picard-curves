```python
R.<x0,x1,x2> = PolynomialRing(ZZ, 3)
```

```python
disc_poly = 256*x0^3 - 128*x0^2*x2^2 + 144*x0*x1^2*x2 + 16*x0*x2^4-27*x1^4-4*x1^2*x2^3
```

```python
S.<a1,b1,a2,b2> = PolynomialRing(ZZ,4)
```

```python
zw_re = a1*a2-b1*b2
zw_im = a1*b2-b1*b2+a2*b1
```

```python
z2re = zw_re.subs(a2=a1, b2=b1)
z2im = zw_im.subs(a2=a1, b2=b1) ## z^2
```

```python
z3re = zw_re.subs(a2=z2re, b2=z2im)
z3im = zw_im.subs(a2=z2re, b2=z2im)   ## z^3
print(z3re,z3im)
```

```python
z4re = zw_re.subs(a2=z3re, b2=z3im)
z4im = zw_im.subs(a2=z3re, b2=z3im)
print(z4re,z4im)
```

```python
P.<a0,b0,a1,b1,a2,b2> = PolynomialRing(ZZ,6)
```

```python
disc_re = 256*(z3re.subs(a1=a0,b1=b0)) - \
128*(zw_re.subs(a1=z2re.subs(a1=a0,b1=b0), b1=z2im.subs(a1=a0,b1=b0), a2=z2re.subs(a1=a2,b1=b2))) + \
144*(zw_re.subs(a1=z2re, b1= z2im, a2 = zw_re.subs(a1=a0, b1=b0))) + \
16 * (zw_re.subs(a1=a0, b1=b0, a2 = z4re.subs(a1=a2, b1=b2), b2 = z4im.subs(a1=a2, b1=b2))) - \
27 * z4re -\
4 * (zw_re.subs(a1=z2re, b1=z2im, a2=z3re.subs(a1=a2, b1=b2), b2=z3im.subs(a1=a2, b1=b2)))
print(disc_re)
```

```python
disc_im = 256*(z3im.subs(a1=a0,b1=b0)) - \
128*(zw_im.subs(a1=z2re.subs(a1=a0,b1=b0), b1=z2im.subs(a1=a0,b1=b0), a2=z2re.subs(a1=a2,b1=b2))) + \
144*(zw_im.subs(a1=z2re, b1= z2im, a2 = zw_re.subs(a1=a0, b1=b0))) + \
16 * (zw_im.subs(a1=a0, b1=b0, a2 = z4re.subs(a1=a2, b1=b2), b2 = z4im.subs(a1=a2, b1=b2))) - \
27 * z4im -\
4 * (zw_im.subs(a1=z2re, b1=z2im, a2=z3re.subs(a1=a2, b1=b2), b2=z3im.subs(a1=a2, b1=b2)))
print(disc_im)
```

```python
norm_poly = a1^2 - a1*b1 + b1^2
norm_of_disc = norm_poly.subs(a1=disc_re, b1=disc_im)
print(norm_of_disc)
```
